```sql
%python
dbutils.widgets.removeAll()
```

```sql
%python
dbutils.widgets.text("nameContainer","unit-catalog")
dbutils.widgets.text("nameStorage","adlsproysmartdata01")
dbutils.widgets.text("catalogo","catalog_au")
```

```sql
%python

nameContainer = dbutils.widgets.get("nameContainer")
nameStorage = dbutils.widgets.get("nameStorage")
catalogo = dbutils.widgets.get("catalogo")

ruta = f"abfss://{nameContainer}@{nameStorage}.dfs.core.windows.net"
```

### Eliminacion tablas Bronze

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.bronze.sms_antiguo;
DROP TABLE IF EXISTS ${catalogo}.bronze.sms_nuevo;
DROP TABLE IF EXISTS ${catalogo}.bronze.catalogo_cliente_producto;
DROP TABLE IF EXISTS ${catalogo}.bronze.emails;
```

```sql
%python
## REMOVE DATA (Bronze)
dbutils.fs.rm(f"{ruta}/tablas/sms_antiguo", True)
dbutils.fs.rm(f"{ruta}/tablas/sms_nuevo", True)
dbutils.fs.rm(f"{ruta}/tablas/catalogo_cliente_producto", True)
dbutils.fs.rm(f"{ruta}/tablas/emails", True)
```

### Eliminacion tablas Silver

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.silver.catalogo_transformed;
DROP TABLE IF EXISTS ${catalogo}.silver.sms_transformed;
DROP TABLE IF EXISTS ${catalogo}.silver.emails_transformed;
```

```sql
%python
# REMOVE DATA (Silver)
dbutils.fs.rm(f"{ruta}/tablas/catalogo_transformed", True)
dbutils.fs.rm(f"{ruta}/tablas/sms_transformed", True)
dbutils.fs.rm(f"{ruta}/tablas/emails_transformed", True)
```

### Eliminacion tablas Golden

```sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.golden.golden_email;
DROP TABLE IF EXISTS ${catalogo}.golden.golden_sms;
```

```sql
%python
# REMOVE DATA (Golden)
dbutils.fs.rm(f"{ruta}/tablas/golden_email", True)
dbutils.fs.rm(f"{ruta}/tablas/golden_sms", True)
```
